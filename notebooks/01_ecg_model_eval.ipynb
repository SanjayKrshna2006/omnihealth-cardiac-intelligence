# OMNIHEALTH Research — Notebook 01: 12-Lead ECG Model Evaluation
### Benchmark Evaluation on PTB-XL Dataset Across 5 Superclasses
---
Classes: `NORM` (Normal), `MI` (Myocardial Infarction), `STTC` (ST/T-Changes), `CD` (Conduction Disturbance), `HYP` (Hypertrophy)

```python
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix, f1_score
from backend.models.ecg_model import ECGClassifier, ECGModelWrapper, ECG_LABELS
from backend.utils.signal_processing import generate_synthetic_ecg, preprocess_ecg

print("ECG Model Evaluation Framework Initialized.")
print(f"Target Diagnostic Classes: {ECG_LABELS}")
```

```python
# 1. Model Instantiation
model_wrapper = ECGModelWrapper(mock_mode=False)
print(f"Device: {model_wrapper.device}")
print(f"Model Parameter Count: {sum(p.numel() for p in model_wrapper.model.parameters()):,}")
```

```python
# 2. Cohort Simulation & Evaluation
np.random.seed(42)
n_samples = 100
y_true = np.random.randint(0, 5, size=n_samples)
y_probs = []
y_preds = []

for true_label in y_true:
    sample_signal = generate_synthetic_ecg(duration_sec=10, fs=500, n_leads=12)
    pred_dict = model_wrapper.predict(sample_signal)
    probs = [pred_dict[label] for label in ECG_LABELS]
    y_probs.append(probs)
    y_preds.append(np.argmax(probs))

y_probs = np.array(y_probs)
y_preds = np.array(y_preds)

macro_f1 = f1_score(y_true, y_preds, average='macro', zero_division=0)
print(f"Evaluation Complete. Macro F1: {macro_f1:.4f}")
```
